# Show result tables

Reads the CSV tables produced by `make_tables_notebooks.py` (in `results/tables_check/`) and displays them.
Numbers are **gross** (no transaction costs, execution look-ahead not yet fixed) — see the publication plan.

```python
import pandas as pd, glob, os
pd.set_option('display.width', 220)
pd.set_option('display.max_columns', 30)

TABLES_DIR = 'results/tables_check'   # output folder of make_tables_notebooks.py

def show(config='12_26_9', scenario='simulation', system='System2', tf='1_hour'):
    """Load and return one table. config e.g. '12_26_9', scenario one_pos/simulation,
    system System1/2/3, tf 1_day/1_hour/5_minute."""
    path = f'{TABLES_DIR}/{config}_{scenario}_{system}_{tf}.csv'
    return pd.read_csv(path, index_col=[0, 1])

def available():
    """List which (config, scenario, system, tf) tables exist."""
    return sorted(os.path.basename(f)[:-4] for f in glob.glob(f'{TABLES_DIR}/*.csv'))

print(len(available()), 'tables available')
```

## Headline table — 12_26_9, simulation, System 2, 1-hour

```python
show('12_26_9', 'simulation', 'System2', '1_hour')
```

## Same config, 5-minute

```python
show('12_26_9', 'simulation', 'System2', '5_minute')
```

## Rules-vs-MACD summary across every table (gross Average Return)

```python
rows = []
for f in sorted(glob.glob(f'{TABLES_DIR}/*.csv')):
    df = pd.read_csv(f, index_col=[0, 1])
    tf = '1_day' if '1_day' in f else ('1_hour' if '1_hour' in f else '5_minute')
    for strat in df.index.get_level_values(0).unique():
        sub = df.loc[strat]
        macd = sub.loc['MACD', 'Average Return']
        best = sub.loc[['Rule 1', 'Rule 2', 'Rule 3', 'GA'], 'Average Return'].max()
        rows.append({'timeframe': tf, 'strategy': strat, 'rule_beats_macd': best >= macd})
summary = pd.DataFrame(rows)
summary.groupby('timeframe')['rule_beats_macd'].mean().mul(100).round(1)
```
